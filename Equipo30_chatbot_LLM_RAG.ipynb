**Introducción a la problemática de la educación financiera**

**Problemática de la Educación Financiera:**

La falta de educación financiera es un problema global con consecuencias significativas a nivel individual y social. Las personas con bajos niveles de conocimiento financiero a menudo toman decisiones subóptimas sobre el ahorro, la inversión, el manejo de deudas y la planificación para el futuro. Esto puede llevar a:

1.  **Inestabilidad Financiera Personal:** Dificultad para cubrir gastos básicos, endeudamiento excesivo, incapacidad para ahorrar para emergencias o metas a largo plazo (como la jubilación o la educación).


2.  **Exclusión Financiera:** Barreras para acceder a productos y servicios financieros que podrían ser beneficiosos.


3.  **Vulnerabilidad a Fraudes:** Mayor susceptibilidad a estafas y productos financieros engañosos.


4.  **Impacto en el Bienestar General:** El estrés financiero puede afectar la salud mental y física, así como las relaciones personales.


5.  **Desigualdad Económica:** La brecha en la educación financiera contribuye a perpetuar la desigualdad.

A menudo, la educación financiera formal es limitada o inexistente en los planes de estudio tradicionales, y la información disponible puede ser compleja, intimidante o poco accesible para el público en general.

**Cómo un Chatbot puede Ayudar a Mejorar la Educación Financiera:**


Un chatbot basado en LLM (Large Language Model) con un sistema RAG (Retrieval Augmented Generation) puede ser una herramienta poderosa para abordar esta problemática de varias maneras:

1.  **Acceso 24/7 y Escalabilidad:** Los usuarios pueden acceder a información y orientación financiera en cualquier momento y desde cualquier lugar, superando las limitaciones de horarios o la disponibilidad de asesores humanos.

2.  **Personalización y Adaptabilidad:** Un chatbot puede adaptar sus respuestas al nivel de conocimiento y a las preguntas específicas del usuario, haciendo que la información sea más comprensible y relevante.

3.  **Información Clara y Sencilla:** Puede simplificar conceptos financieros complejos y explicarlos en lenguaje llano, rompiendo la barrera de la jerga técnica.

4.  **Respuesta a Preguntas Frecuentes:** Puede manejar eficientemente un gran volumen de preguntas comunes sobre ahorro, presupuesto, crédito, inversiones básicas, etc.

5.  **Generación de Ejemplos y Escenarios:** Un chatbot puede crear ejemplos prácticos y escenarios hipotéticos para ilustrar conceptos financieros y sus consecuencias.

6.  **Fomento de la Interacción:** La naturaleza conversacional de un chatbot puede hacer que el aprendizaje sobre finanzas sea menos aburrido y más interactivo que leer materiales estáticos.

7.  **Base de Conocimiento Actualizada (RAG):** Al integrar un sistema RAG, el chatbot puede recuperar información relevante y actualizada de fuentes confiables (como documentos PDF, sitios web oficiales, etc.), asegurando que las respuestas sean precisas y estén basadas en datos verificados, lo cual es crucial en un área como las finanzas donde la información cambia.

8.  **Reducción de Costos:** Proporcionar educación financiera a través de un chatbot es potencialmente más económico y escalable que la asesoría individual o talleres presenciales masivos.

9.  **Primer Punto de Contacto Seguro:** Ofrece un entorno anónimo donde los usuarios pueden hacer preguntas sobre temas financieros personales sin sentirse juzgados.


En resumen, un chatbot con capacidades RAG puede hacer mas acequible el acceso a la educación financiera, proporcionando información personalizada, accesible y confiable, lo que permite a las personas tomar la rienda de sus finanzas para tomar mejores decisiones financieras y mejorar su bienestar económico.


**Reporte Técnico del Sistema RAG + LLM**

Básicamente es un modelo de lenguaje grande que puede hacer uso de información propia (LLM), por medio del la generación aumentada por recuperación (RAG) Básicamente aumenta la calidad de las respuesta por medio de la actualización de fuentes externas sobre modelos ya compilados, esto es una gran ventaja porque se pueden actualizar o moldear las respuestas que se necesitan para ciertos contexto.

Como se sabe el RAG tiene 2 pasos:

* Recuperación

* Generación

Nosotros para el proceso de la generación de un sistema externo y su recuperación utilizamos el modelo "**embedding-001**" con Google API

Y para no volver complejo el análsiis para el proceso de generación "**gemini-1.5-flash-latest**" 2 razones específicas nos hicieron tomar estos modelos aunque se hicieron pruebas con muchos, 1 es que son gratuitos con acceso amplio y la otra es que fueron los más veloces como el chatbot o asistente es en tiempo real esto adquiere una relevancia importante.

Para el embedding model se utilizó la siguiente instrucción:

genai.embed_content(model=embedding_model,
                             content=text,
                             task_type="retrieval_document",
                             title=title)["embedding"]

Parámetros
**model** = modelo a utilizar en este caso **embedding-001** (su propósito es la simplificación de los tokens para poder ser procesados correctamente y ser utilizados como una fuente externa al modelo de generación)
**context** = texto a vectorizar
**task_type** = Pueden ser diferentes tipos:

Task type	Description
CLASSIFICATION	Optimizada para clasificación de textos

CLUSTERING	Optimizada para textos clusterizados para categorizarlo con base en sus similitudes

**RETRIEVAL_DOCUMENT**, RETRIEVAL_QUERY, QUESTION_ANSWERING, and FACT_VERIFICATION	Optimizado para la obtención de información

CODE_RETRIEVAL_QUERY	Obtiene información de estructuras previamente generadas con RETRIEVAL_DOCUMENT pero con estructuras de datos complejas.

SEMANTIC_SIMILARITY	Optimizado para buscar similitudes en el texto pero no para la obtención de información.

title= Título del documento



Para el modelo de Generacións obtamos por **gemini-1.5-flash-latest**

La instrucción que se uso fue:

model_generation = genai.GenerativeModel(generation_model)

**Supported inputs & outputs**
Inputs: Text, Code, Images, Audio, Video

Outputs: Text

**Token limits**
Maximum input tokens: 1,048,576

Maximum output tokens: 8,192 (default)

```python
df1 = pd.DataFrame(documents)
df1.columns = ['Title', 'Content']
df2 = pd.DataFrame([{
    'Title': doc.metadata['source'],
    'Content': doc.page_content} for doc in docs])
df = pd.concat([df1, df2])
print(df2)
df = pd.concat([df1, df2])
print(df2)
```

```python
df.dropna(subset=['Content'], inplace=True)
df = df[df['Content'].str.strip() != '']
```

```python
# Agreguemos los vectores embebidos en una tercer columna:

def embed_fn(title, text):
  return genai.embed_content(model=embedding_model,
                             content=text,
                             task_type="retrieval_document",
                             title=title)["embedding"]

# Filter out rows with empty or whitespace-only content before embedding
#df = df[df['Content'].str.strip() != '']

df['Embeddings'] = df.apply(lambda row: embed_fn(row['Title'], row['Content']), axis=1)
```

```python
# Create a vector store (using ChromaDB)
# You can specify a directory to persist the vector store
persist_directory = 'db'
# Check if the persist directory exists and load from it, otherwise create a new one
if os.path.exists(persist_directory):
    # Instantiate the embedding function when loading from persistence
    embeddings = GoogleGenerativeAIEmbeddings(model=embedding_model, google_api_key=GOOGLE_API_KEY)
    vectordb = Chroma(persist_directory=persist_directory, embedding_function=embeddings)
    print(f"Loaded existing vector store from {persist_directory}")
else:
    print(f"Creating new vector store in {persist_directory}")
    # Instantiate the embedding function for creating the vector store
    embeddings = GoogleGenerativeAIEmbeddings(model=embedding_model, google_api_key=GOOGLE_API_KEY)
    # Use Chroma.from_documents with the list of Document objects and the embedding function
    vectordb = Chroma.from_documents(documents=docs, embedding=embeddings, persist_directory=persist_directory)
    vectordb.persist()
    print("Vector store created and persisted.")
```

```python
def find_best_passage(query, dataframe):
  """
  Cálcula la distancia entre el query y cada documento en el dataframe usando el punto producto.
  """
  query_embedding = genai.embed_content(model=embedding_model,
                                        content=query,
                                        task_type="retrieval_query")
  dot_products = np.dot(np.stack(dataframe['Embeddings']), query_embedding["embedding"])
  idx = np.argmax(dot_products)
  return dataframe.iloc[idx]['Content'] # Return text from index with max value
```

```python
def chat_with_rag(query):

  prompt = textwrap.dedent("""Eres un útil e informativo asistente que responde preguntas usando texto de referencia del pasage incluido a continuación. \
  Asegurate de responder oraciones completas, ser comprensivo, incluir toda la información contextual relevante. \
  Sin embargo, estas hablando a una audiencia no técnica y entrenada, por tanto asegurate de desmenusar argumentos complicados y \
  utilizar un tone amigable y simple. \
  Si el pasaje es irrelevante para la respuesta, puedes ignorarlo.

  Pregunta: '{query}'
    Respuesta:
  """).format(query=query)

  return prompt

print("Chatbot initialized. Type 'exit' to quit.")

# Punto de RUBRICA La interfaz del chat es simple y clara sin pretenciones y dedicada exclusivamente a la pregunta y respuesta.
while True:
    user_input = input("Listo: ")
    if user_input.lower() == 'exit':
        break
    if not user_input.strip():
        print("Bot: Estoy atento a sus preguntas.")
        continue

#answer = model_generation.generate_content(prompt)
    response_temp = model_generation.generate_content(chat_with_rag(user_input))
    bot_response = response_temp.text
    print(f"Meliha: {bot_response}")

print("Chat session ended.")
```
